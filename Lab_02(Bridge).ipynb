```python
#___________________TASK-2______________________

# Use the actual subject columns present in your CSV
subjects = ['Math', 'Stats', 'Python', 'Databases', 'English']

# Calculate average marks for each student
df['Average'] = df[subjects].mean(axis=1)

# Summarize average marks for each section
summary = df.groupby('Section')['Average'].mean().round(2)

# Display the summary
print(summary)
```

```python
# Define the actual subject columns from the dataset
subjects = ['Math', 'Stats', 'Python', 'Databases', 'English']

# Add Average column for every student
df['Average'] = df[subjects].mean(axis=1).round(2)

# Print the top 3 students with their section (using 'Name' instead of 'Student')
top3 = df.nlargest(3, 'Average')[['Name', 'Section', 'Average']]

print("Top 3 Students:")
print(top3.to_string(index=False))
```

```python
# Calculate the average marks for each section
section_avg = df.groupby('Section')['Average'].mean().round(2)

# Find the section with the highest average
highest_section = section_avg.idxmax()
highest_average = section_avg.max()

# Print the result
print("Section with highest average:", highest_section)
print("Highest average marks:", highest_average)
```

```python
#__________________TAsk-03___________________

import json

# Add Average column if not already created (using the correct subject columns)
subjects = ['Math', 'Stats', 'Python', 'Databases', 'English']
df['Average'] = df[subjects].mean(axis=1).round(2)

# Create summary: average marks for each section
summary = df.groupby('Section', as_index=False)['Average'].mean()
summary['Average'] = summary['Average'].round(2)

# Save summary as summary.json
summary.to_json('summary.json', orient='records', indent=4)

# Save full dataset as students.json
df.to_json('students.json', orient='records', indent=4)

print("Files saved successfully!")
```

```python
#_______________TASK-04____________________

%%writefile t4.js
const fs = require('fs');

const students = JSON.parse(fs.readFileSync('students.json', 'utf8'));
const summary = JSON.parse(fs.readFileSync('summary.json', 'utf8'));

console.log('Number of students:', students.length);

const below60 = students
    .filter(student => student.Average < 60)
    .map(student => student.Student);

console.log('Students with Average below 60:', below60);

const highestSection = [...summary].sort(
    (a, b) => b.Average - a.Average
)[0];

console.log(
    'Section with highest average:',
    highestSection.Section,
    'Average:',
    highestSection.Average
);

summary.forEach(section => {
    console.log(`Section ${section.Section}: ${section.Average}`);
});
```

```python
#_______________TASK-05________________

%%writefile chart.js
const fs = require('fs');

const summary = JSON.parse(
    fs.readFileSync('summary.json', 'utf8')
);

// Labels and values for the chart
const sections = summary.map(item => `Section ${item.Section}`);
const averages = summary.map(item => item.Average);

const chartHTML = `
<!DOCTYPE html>
<html>
<head>
    <title>Average Python Marks by Section</title>
    <script src="https://cdn.jsdelivr.net/npm/chart.js"></script>
</head>
<body>
    <h2 style="text-align:center">Average Python Marks by Section</h2>

    <div style="width:80%; max-width:900px; margin:auto;">
        <canvas id="myChart"></canvas>
    </div>

    <script>
        const sections = ${JSON.stringify(sections)};
        const averages = ${JSON.stringify(averages)};

        new Chart(document.getElementById('myChart'), {
            type: 'bar',
            data: {
                labels: sections,
                datasets: [{
                    label: 'Average Python Marks',
                    data: averages,
                    backgroundColor: 'steelblue',
                    borderColor: 'navy',
                    borderWidth: 1
                }]
            },
            options: {
                responsive: true,
                plugins: {
                    title: {
                        display: true,
                        text: 'Average Python Marks by Section'
                    },
                    datalabels: {
                        display: false
                    }
                },
                scales: {
                    y: {
                        beginAtZero: true,
                        title: {
                            display: true,
                            text: 'Average Marks'
                        }
                    },
                    x: {
                        title: {
                            display: true,
                            text: 'Section'
                        }
                    }
                }
            },
            plugins: [{
                id: 'valueLabels',
                afterDatasetsDraw(chart) {
                    const ctx = chart.ctx;
                    chart.data.datasets.forEach((dataset, i) => {
                        const meta = chart.getDatasetMeta(i);
                        meta.data.forEach((bar, index) => {
                            ctx.fillStyle = 'black';
                            ctx.textAlign = 'center';
                            ctx.fillText(
                                dataset.data[index],
                                bar.x,
                                bar.y - 5
                            );
                        });
                    });
                }
            }]
        });
    </script>
</body>
</html>
`;

fs.writeFileSync('chart.html', chartHTML);
console.log('chart.html created successfully!');
```
